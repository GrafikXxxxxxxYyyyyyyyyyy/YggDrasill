### Быстрый запуск в 2 строчки кода

```python
from yggdrasill import Hypergraph 

graph = Hypergraph.from_template("sdxl_text2img", device="cuda")

output = graph.run(
    prompt="A photo of an astronaut riding a horse on mars",
    num_inference_steps=50,
    guidance_scale=7.5,
    seed=42,
    width=1024,
    height=1024,
)

img = output.images[0] if output.images else None
img
```

### Сборка своего графа

```python
from yggdrasill import Hypergraph

graph = Hypergraph(name="MyAwesomeGraph")

pretrained = "stabilityai/stable-diffusion-xl-base-1.0"

# Добавление компонентов графа с автосвязыванием
# Внутренние служебные ноды будут встроены автоматически под капотом
graph.add_node("MyAwesomePromptEncoder", type="sdxl.prompt_encoder", pretrained=pretrained)
graph.add_node("MyAwesomeBackbone", type="sdxl.backbone", pretrained=pretrained)
graph.add_node("MyAwesomeScheduler", type="sdxl.scheduler", pretrained=pretrained)
graph.add_node("MyAwesomeAutoencoder", type="sdxl.autoencoder", pretrained=pretrained)

graph.infer_exposed_ports()
graph.to("cuda")
```

### Замена компонентов в готовом графе

```python
graph = Hypergraph.from_template("sdxl_text2img", device="cuda")

new_repo = "stabilityai/stable-diffusion-xl-base-1.0"

# Замена основных компонентов по понятным пользовательским именам
# Внутренние узлы вроде scheduler setup/step скрыты за alias
graph.replace_node("prompt_encoder", pretrained=new_repo)
graph.replace_node("backbone", pretrained=new_repo)
graph.replace_node("scheduler", pretrained=new_repo)
graph.replace_node("autoencoder", pretrained=new_repo)
```

### Добавление ControlNet + IP-Adapter к готовому графу

```python
graph = Hypergraph.from_template("sdxl_text2img", device="cuda")

# Добавление ControlNet — автоматически подключается к backbone
graph.add_node("MyAwesomeControlNet", type="adapter.controlnet", pretrained="lllyasviel/sd-controlnet-canny")

# Добавление IP-Adapter — автоматически подключается к backbone
graph.add_node("MyAwesomeIPAdapter", type="adapter.ip_adapter", pretrained="h94/IP-Adapter")
```

### Добавление разных ControlNet и разных IP-adapter и сборка полного графа с ними

```python
graph = Hypergraph("MyAwesomeGraph")

pretrained = "stabilityai/stable-diffusion-xl-base-1.0"

# Собираем SDXL граф из крупных пользовательских компонентов
# Служебные ноды будут встроены автоматически под капотом
graph.add_node("MyAwesomePromptEncoder", type="sdxl.prompt_encoder", pretrained=pretrained)
graph.add_node("MyAwesomeBackbone", type="sdxl.backbone", pretrained=pretrained)
graph.add_node("MyAwesomeScheduler", type="sdxl.scheduler", pretrained=pretrained)
graph.add_node("MyAwesomeAutoencoder", type="sdxl.autoencoder", pretrained=pretrained)

# Добавляем ControlNet'ы — каждый автоматически подключается к backbone
graph.add_node("MyAwesomeControlNet1", type="adapter.controlnet", pretrained="lllyasviel/sd-controlnet-canny")
graph.add_node("MyAwesomeControlNet2", type="adapter.controlnet", pretrained="lllyasviel/sd-controlnet-depth")

# Добавляем IP-Adapter'ы — каждый автоматически подключается к backbone
graph.add_node("MyAwesomeIPAdapter1", type="adapter.ip_adapter", pretrained="h94/IP-Adapter")
graph.add_node("MyAwesomeIPAdapter2", type="adapter.ip_adapter", pretrained="h94/IP-Adapter")

graph.infer_exposed_ports()
graph.to("cuda")
```
